```python
from sklearn.base import BaseEstimator

class MultiplyBy(BaseEstimator):
    def __init__(self, factor=2):
        self.factor = factor
    
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        return X * self.factor

mulby3 = MultiplyBy(factor = 3)
X = 21
mulby3.fit(X)
mulby3.transform(X)
```

```python
mulby3.get_params()
```

```python
mulby3.set_params(factor=12)
mulby3.transform(X)
```

```python
class DummyClassifier(BaseEstimator):
    def __init__(self, length=12):
        self.length = length
    
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        return X
    
    def predict(self, X):
        return [X for _ in range(self.length)]  

dc = DummyClassifier(length=10)
X = 99
dc.fit(X)
print(dc.predict(X))
```
