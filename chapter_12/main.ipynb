# Deep Computer Vision using Convotional Neural Network


```python
import numpy as np
import torch
from sklearn.datasets import load_sample_images
```

```python
sample_images = np.stack(load_sample_images()['images'])
sample_images = torch.tensor(sample_images, dtype=torch.float32) / 255
```

```python
sample_images.shape
```

```python
sample_images_permuted = sample_images.permute(0, 3, 1, 2)
sample_images_permuted.shape
```

```python
import torchvision
import torchvision.transforms.v2 as T

cropped_images = T.CenterCrop((70, 120))(sample_images_permuted)
cropped_images.shape
```

## Convtional layer

```python
import torch.nn as nn

torch.manual_seed(122)
conv_layer = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=7)
# conv_layer = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=7, padding='same')
# conv_layer = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=7, padding=3, stride=(2,2))
fmap = conv_layer(cropped_images)
```

```python
fmap.shape
```

```python
# weight --> [output_channels, input_channels, kernel_height, kernel_width], bias --> [output_channels]
conv_layer.weight.shape, conv_layer.bias.shape
```

It’s important to add an activation function after each
convolutional layer. This is for the same reason as for nn.Linear
layers: a convolutional layer performs a linear operation, so if you
stacked multiple convolutional layers without any activation
functions, they would all be equivalent to a single convolutional
layer, and they wouldn’t be able to learn anything really complex.

## Pooling layer

```python
max_pool = nn.MaxPool2d(kernel_size=2,stride=2)
mxp = max_pool(fmap)
```

```python
mxp.shape
```

```python
global_avg_pool = nn.AdaptiveAvgPool2d(output_size=1)
output = global_avg_pool(cropped_images)
```
